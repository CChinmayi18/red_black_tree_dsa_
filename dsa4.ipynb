```python

# Kruskal's and Prim's Algorithm

# 1. KRUSKAL'S ALGORITHM

def kruskal(vertices, edges):

    parent = list(range(vertices))

    def find(v):
        if parent[v] != v:
            parent[v] = find(parent[v])
        return parent[v]

    def union(u, v):
        root_u = find(u)
        root_v = find(v)

        if root_u != root_v:
            parent[root_v] = root_u
            return True
        return False

    mst = []
    total_weight = 0

    sorted_edges = sorted(edges, key=lambda x: x[2])

    for u, v, weight in sorted_edges:
        if union(u, v):
            mst.append((u, v))
            total_weight += weight

        if len(mst) == vertices - 1:
            break

    if len(mst) != vertices - 1:
        return None, None

    return mst, total_weight


# 2. PRIM'S ALGORITHM

def prim(vertices, edges):

    if vertices == 0:
        return None, None

    graph = [[] for _ in range(vertices)]

    for u, v, weight in edges:
        graph[u].append((v, weight))
        graph[v].append((u, weight))

    visited = [False] * vertices
    mst = []
    total_weight = 0

    visited[0] = True
    visit_order = [0]

    while len(mst) < vertices - 1:

        min_edge = None

        for u in visit_order:
            for v, weight in graph[u]:

                if not visited[v]:

                    if (min_edge is None or
                        weight < min_edge[2] or
                        (weight == min_edge[2] and
                         u < min_edge[0])):

                        min_edge = (u, v, weight)

        if min_edge is None:
            return None, None

        u, v, weight = min_edge

        mst.append((u, v))
        total_weight += weight

        visited[v] = True
        visit_order.append(v)

    return mst, total_weight


# 3. DISPLAY FUNCTION

def display_mst(name, mst, weight):

    print("\n" + name)
    print("Minimum Spanning Tree:")

    for u, v in mst:
        print(f"{u + 1} -> {v + 1}")

    print("Total Weight:", weight)


# 4. TEST CASES

test_cases = [

    # Test Case 1
    (0, []),

    # Test Case 2
    (4, [(2, 3, 2), (0, 3, 2),
         (0, 1, 1), (1, 2, 1)]),

    # Test Case 3
    (4, [(0, 1, 1), (1, 2, 2),
         (2, 3, 3), (0, 3, 10),
         (0, 2, 8)]),

    # Test Case 4
    (4, [(0, 1, 5), (1, 2, -3),
         (2, 3, 4), (0, 3, 6)])
]


# 5. EXECUTION

for i, (vertices, edges) in enumerate(test_cases, 1):

    print("\nTEST CASE", i)
    print("Vertices:", vertices)
    print("Edges:", edges)

    if vertices == 0:
        print("Output: -1")
        continue

    if any(weight < 0 for u, v, weight in edges):
        print("Output: Invalid - Negative weights not allowed")
        continue

    k_mst, k_weight = kruskal(vertices, edges)
    p_mst, p_weight = prim(vertices, edges)

    if k_mst is None or p_mst is None:
        print("Output: Minimum Spanning Tree not possible")
        continue

    display_mst("Kruskal's Algorithm", k_mst, k_weight)

    display_mst("Prim's Algorithm", p_mst, p_weight)

    if set(k_mst) == set(p_mst):
        print("\nBoth algorithms produce SAME MST")
    else:
        print("\nBoth algorithms produce DIFFERENT MSTs")

    print("Same minimum weight:", k_weight == p_weight)
```
