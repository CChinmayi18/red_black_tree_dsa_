```python

# DSA EXPERIMENT 3: HASHING
# Implement hashing using Chaining and Open Addressing.
# Evaluate performance with different hash functions.

import time
import random
import math
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. HASH FUNCTIONS
# --------------------------------------------------

def division_hash(key, size):
    return key % size


def multiplication_hash(key, size):
    A = (math.sqrt(5) - 1) / 2
    return int(size * ((key * A) % 1))


def mid_square_hash(key, size):
    square = str(key * key)
    digits = len(str(size - 1))

    # Add zeros if the square has fewer digits
    square = square.zfill(digits + 2)

    middle = len(square) // 2
    start = max(0, middle - digits // 2)
    middle_digits = square[start:start + digits]

    return int(middle_digits) % size


hash_functions = {
    "Division": division_hash,
    "Multiplication": multiplication_hash,
    "Mid-Square": mid_square_hash
}

# --------------------------------------------------
# 2. SEPARATE CHAINING
# --------------------------------------------------

class ChainingHashTable:
    def __init__(self, size, hash_function):
        self.size = size
        self.table = [[] for _ in range(size)]
        self.hash_function = hash_function
        self.collisions = 0

    def insert(self, key):
        index = self.hash_function(key, self.size)

        if key in self.table[index]:
            return

        if len(self.table[index]) > 0:
            self.collisions += 1

        self.table[index].append(key)

    def search(self, key):
        index = self.hash_function(key, self.size)
        return key in self.table[index]


# --------------------------------------------------
# 3. OPEN ADDRESSING
# --------------------------------------------------

class OpenAddressingHashTable:
    def __init__(self, size, hash_function, method):
        self.size = size
        self.table = [None] * size
        self.hash_function = hash_function
        self.method = method
        self.collisions = 0

    def get_index(self, key, i):
        h1 = self.hash_function(key, self.size)

        if self.method == "Linear Probing":
            return (h1 + i) % self.size

        elif self.method == "Quadratic Probing":
            return (h1 + i * i) % self.size

        elif self.method == "Double Hashing":
            # The table size is prime, so this step
            # is nonzero and relatively prime to size.
            h2 = 1 + (key % (self.size - 1))
            return (h1 + i * h2) % self.size

    def insert(self, key):
        for i in range(self.size):
            index = self.get_index(key, i)

            if self.table[index] == key:
                return

            if self.table[index] is None:
                self.table[index] = key
                return

            self.collisions += 1

        raise OverflowError(
            "Hash table is full. Cannot insert key: " + str(key)
        )

    def search(self, key):
        for i in range(self.size):
            index = self.get_index(key, i)

            if self.table[index] is None:
                return False

            if self.table[index] == key:
                return True

        return False


# --------------------------------------------------
# 4. PERFORMANCE EVALUATION
# --------------------------------------------------

def evaluate_hashing(keys, search_keys, table_size):
    results = []

    for hash_name, hash_function in hash_functions.items():

        # Evaluate Separate Chaining
        chaining = ChainingHashTable(table_size, hash_function)

        start = time.perf_counter()
        for key in keys:
            chaining.insert(key)
        insertion_time = time.perf_counter() - start

        start = time.perf_counter()
        for key in search_keys:
            chaining.search(key)
        search_time = time.perf_counter() - start

        results.append({
            "Hash Function": hash_name,
            "Collision Method": "Chaining",
            "Collisions": chaining.collisions,
            "Insertion Time (ms)": insertion_time * 1000,
            "Search Time (ms)": search_time * 1000,
            "Load Factor": round(len(keys) / table_size, 3)
        })

        # Evaluate Linear Probing, Quadratic Probing,
        # and Double Hashing
        for method in [
            "Linear Probing",
            "Quadratic Probing",
            "Double Hashing"
        ]:
            table = OpenAddressingHashTable(
                table_size, hash_function, method
            )

            start = time.perf_counter()
            for key in keys:
                table.insert(key)
            insertion_time = time.perf_counter() - start

            start = time.perf_counter()
            for key in search_keys:
                table.search(key)
            search_time = time.perf_counter() - start

            results.append({
                "Hash Function": hash_name,
                "Collision Method": method,
                "Collisions": table.collisions,
                "Insertion Time (ms)": insertion_time * 1000,
                "Search Time (ms)": search_time * 1000,
                "Load Factor": round(len(keys) / table_size, 3)
            })

    return pd.DataFrame(results)


# --------------------------------------------------
# 5. MAIN PROGRAM
# --------------------------------------------------

random.seed(42)

TABLE_SIZE = 1009  # Prime number
NUMBER_OF_KEYS = 500

# Generate unique positive integer keys
keys = random.sample(range(10000, 1000000), NUMBER_OF_KEYS)

# Search for inserted keys
search_keys = keys[:100]

results = evaluate_hashing(
    keys, search_keys, TABLE_SIZE
)

# --------------------------------------------------
# 6. DISPLAY RESULTS
# --------------------------------------------------

print("DSA EXPERIMENT 3: HASHING")
print("=" * 75)
print("Number of keys:", NUMBER_OF_KEYS)
print("Hash table size:", TABLE_SIZE)
print("Search operations:", len(search_keys))
print("Load factor:", round(NUMBER_OF_KEYS / TABLE_SIZE, 3))

print("\nPERFORMANCE COMPARISON")
print("=" * 75)

print(results.round({
    "Insertion Time (ms)": 4,
    "Search Time (ms)": 4
}).to_string(index=False))

# --------------------------------------------------
# 7. PLOT COLLISION COMPARISON
# --------------------------------------------------

pivot_collisions = results.pivot(
    index="Hash Function",
    columns="Collision Method",
    values="Collisions"
)

pivot_collisions.plot(kind="bar", figsize=(12, 6))

plt.title("Collision Comparison of Hashing Techniques")
plt.xlabel("Hash Function")
plt.ylabel("Number of Collisions")
plt.xticks(rotation=0)
plt.legend(title="Collision Method")
plt.tight_layout()
plt.show()

# --------------------------------------------------
# 8. PLOT INSERTION TIME COMPARISON
# --------------------------------------------------

pivot_time = results.pivot(
    index="Hash Function",
    columns="Collision Method",
    values="Insertion Time (ms)"
)

pivot_time.plot(kind="bar", figsize=(12, 6))

plt.title("Insertion Time Comparison of Hashing Techniques")
plt.xlabel("Hash Function")
plt.ylabel("Insertion Time (milliseconds)")
plt.xticks(rotation=0)
plt.legend(title="Collision Method")
plt.tight_layout()
plt.show()

# --------------------------------------------------
# 9. SEARCH CORRECTNESS CHECK
# --------------------------------------------------

print("\nSEARCH CORRECTNESS CHECK")
print("=" * 75)

all_found = True

for hash_name, hash_function in hash_functions.items():
    chaining = ChainingHashTable(TABLE_SIZE, hash_function)
    for key in keys:
        chaining.insert(key)

    if not all(chaining.search(key) for key in search_keys):
        all_found = False

    for method in [
        "Linear Probing",
        "Quadratic Probing",
        "Double Hashing"
    ]:
        table = OpenAddressingHashTable(
            TABLE_SIZE, hash_function, method
        )

        for key in keys:
            table.insert(key)

        if not all(table.search(key) for key in search_keys):
            all_found = False

print("All inserted search keys found:", all_found)

if all_found:
    print("Experiment completed successfully.")
else:
    print("Some search operations failed. Check the implementation.")
```
